```python
print("Simple Calculator")
print("Choose operation: +, -, *, /")

operation = input("Enter operation: ")
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        print("Cannot divide by zero")
        exit()
else:
    print("Invalid operation")
    exit()

print("Result:", result)
```
